## Dataset Check

````python
import os

DATASET_ROOT = '/kaggle/input/datasets/philippstaudinger/project-anna/20251127_rgb_depth_o3de'

for folder in ['rgb', 'depth', 'camera']:
    path = f'{DATASET_ROOT}/{folder}'
    files = os.listdir(path) if os.path.exists(path) else ['MISSING']
    print(f"{folder}/: {len(files)} files — e.g. {files[0]}")
````

## Train

````python
!cd /kaggle/working/apple-vision && MPLBACKEND=agg uv run python -m apple_vision.train_depth \
  --dataset-root /kaggle/input/datasets/philippstaudinger/project-anna/20251127_rgb_depth_o3de \
  --epochs 20 \
  --batch-size 8 \
  --num-workers 4 \
  --resize 640 400 \
  --max-depth 2.5 \
  --out-dir /kaggle/working/checkpoints
````

## Evaluate

````python
!cd /kaggle/working/apple-vision && uv run python -m apple_vision.evaluate_depth \
  --checkpoint /kaggle/working/checkpoints/depth_estimator_best.pth \
  --dataset-root /kaggle/input/datasets/philippstaudinger/project-anna/20251127_rgb_depth_o3de \
  --resize 640 400 \
  --max-depth 2.5
````
